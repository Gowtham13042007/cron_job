```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score

# 1. Load data
diabetes_dataset = pd.read_csv('/content/diabetes.csv')

# 2. Separate features and labels
X = diabetes_dataset.drop(columns='Outcome', axis=1)
Y = diabetes_dataset['Outcome']

# 3. Standardize the features
scaler = StandardScaler()
scaler.fit(X)
X = scaler.transform(X)

# 4. Split into Training and Test sets (80% train, 20% test)
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=2)

# 5. Initialize and Train the SVM Classifier
classifier = svm.SVC(kernel='rbf',C=10,gamma=0.05,max_iter=1000) # small gamma smooth boundary and large gamma complex
#classifier = svm.SVC(
    #kernel='poly',
    #degree=3,
    #C=1
#)
#we can also use class_weight='balanced'
classifier.fit(X_train, Y_train)

# 6. Print Evaluations
print('Training Data Accuracy:', accuracy_score(classifier.predict(X_train), Y_train))
print('Test Data Accuracy:', accuracy_score(classifier.predict(X_test), Y_test))

# 7. Predictive System for a single instance
input_data = (5, 116, 74, 0, 0, 25.6, 0.201, 30)
input_data_reshaped = np.asarray(input_data).reshape(1,-1)
std_data = scaler.transform(input_data_reshaped)

prediction = classifier.predict(std_data)

if prediction[0] == 0:
    print('The person is not diabetic')
else:
    print('The person is diabetic')
```
