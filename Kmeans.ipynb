```python
import random
import numpy as np

def createClusteredData(N,k):
  random.seed(10)
  pointsPerCluster=float(N)/k
  X=[]
  for i in range(k):
    incomeCentriod=random.uniform(20000.0,200000.0)
    ageCentriod=random.uniform(20.0,70.0)
    for j in range(int(pointsPerCluster)):
      X.append([np.random.normal(incomeCentriod,10000.0),np.random.normal(ageCentriod,2.0)])
  X=np.array(X)
  return X

```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from sklearn.preprocessing import scale


data=createClusteredData(100,5)
model=KMeans(n_clusters=5)
model=model.fit(scale(data))  # scaling for mean =0 s.d=1
print(model.labels_)

plt.figure(figsize=(8,6))
plt.scatter(data[:,0],data[:,1],c=model.labels_.astype(float))
plt.show()

```
