```python
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding
from tensorflow.keras.layers import LSTM
from tensorflow.keras.datasets import imdb
```

```python
print('Loading data....')
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=20000)
```

```python
x_train[0]
```

```python
y_train[0]
import pandas as pd

pd.Series(y_train).value_counts()
```

```python
x_train=sequence.pad_sequences(x_train,maxlen=80)
x_test=sequence.pad_sequences(x_test,maxlen=80)  #  it will reduce the token to cutoff or adjust to 80 tokems
```

```python
model=Sequential()
model.add(Embedding(20000,128))  # only 20000 frequent words are kept and each word is represented in the form of 128 array(Words with similar meaning get similar vectors that is what embedding learns)
model.add(LSTM(128,dropout=0.2,recurrent_dropout=0.2)) # 128 neurons and recurrent_dropout means the memory connections dropout
model.add(Dense(1,activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,batch_size=32,epochs=15,verbose=2,validation_data=(x_test,y_test))
```

```python
score,acc==model.evaluate(x_test,y_test,batch_size=32,verbose=2)
print('Test score:',score)
print('Test accuracy:',acc)
```
