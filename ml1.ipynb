```python
import numpy as np
import matplotlib.pyplot as plt

vals=np.random.normal(0,0.5,10000)
plt.hist(vals,50,edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Values')
plt.show()
```

```python
print(np.percentile(vals,50))
print(np.percentile(vals,10))
print(np.percentile(vals,90))
```

```python
print(np.mean(vals))  #first moment
print(np.var(vals)) #second moment
```

```python
import scipy.stats as super
print(super.skew(vals))  #third moment(looks like left side)
print(super.kurtosis(vals))  #fourth moment
```
