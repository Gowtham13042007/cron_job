```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
sns.set()
sns.set(rc={'figure.figsize':(15,5)})
df=pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')

ax = sns.boxplot(x='Pclass', y='Age', hue='Survived', data=df)

ax.set_xticklabels(ax.get_xticklabels(), rotation=90) # just the rotation of the number in x_axis
ax.set_yticklabels(ax.get_yticklabels(), rotation=90) # just the rotation of the number in y_axis

plt.show()
```

```python
ax=sns.swarmplot(x='Pclass',y='Age',hue='Survived',data=df)
ax.set_xticklabels(ax.get_xticklabels(),rotation=90)
ax.set_yticklabels(ax.get_yticklabels(),rotation=90)
plt.show()
```

```python
ax=sns.countplot(x='Pclass',hue='Survived',data=df)   # rememeber we can removve the hue which the data given will always be raw
ax.set_xticklabels(ax.get_xticklabels(),rotation=90)
```

```python
df2=df.pivot_table(index='Survived',columns='Pclass',values='Age',aggfunc='mean')
sns.heatmap(df2)
```
