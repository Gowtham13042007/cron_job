```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(2)
pagespeeds=np.random.normal(3,1,100)
purchaseamount=np.random.normal(5,1,100)/pagespeeds
plt.scatter(pagespeeds,purchaseamount)
plt.show()
```

```python
trainX=pagespeeds[:80]
trainY=purchaseamount[:80]
testX=pagespeeds[80:]
testY=purchaseamount[80:]

plt.scatter(trainX,trainY)
plt.show()
```

```python
plt.scatter(testX,testY)
plt.show()
```

```python
p4=np.poly1d(np.polyfit(trainX,trainY,15))
print(p4)
```

```python
import matplotlib.pyplot as plt

xp=np.linspace(0,7,100)
axes=plt.axes()
axes.set_xlim([0,7])
axes.set_ylim([0,8])
plt.scatter(trainX,trainY)
plt.plot(xp,p4(xp),color='red')
plt.show()
```

```python
xp=np.linspace(0,7,100)
axes=plt.axes()
axes.set_xlim([0,7])
axes.set_ylim([0,8])
plt.scatter(testX,testY)
plt.plot(xp,p4(xp),color='red')
plt.show()
```

```python
from sklearn.metrics import r2_score

r2=r2_score(testY,p4(testX))
print(r2)
```

```python
r2=r2_score(trainY,p4(trainX))
print(r2)
```
